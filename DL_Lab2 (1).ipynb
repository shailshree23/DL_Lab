```python
import numpy as np
```

```python
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
def sigmoid_derivative(x):
  return x*(1-x)
```

```python
x=np.array([[0,0],[0,1],[1,0],[1,1]])
y=np.array([[0],[1],[1],[0]])
```

```python
np.random.seed(0)
```

```python
w1= np.random.uniform(-1, 1, (2,4))
w2= np.random.uniform(-1,1, (4,1))
lr=0.1
```

```python
for epoch in range(20000):
  hidden_layer=sigmoid(x@w1)
  output_layer=sigmoid(hidden_layer@w2)
  error=y-output_layer
  d_output=error*sigmoid_derivative(output_layer)
  d_hidden=(d_output@w2.T)*sigmoid_derivative(hidden_layer)

w2+=hidden_layer.T@d_output*lr
w1+=x.T@d_hidden*lr

if epoch%1000==0:
  print(f"Epoch {epoch}, loss: {np.mean(np.abs(error))}f")

print ("\nfinal output after training:")
print(output_layer.round(3))
```

```python
def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - x**2
```

```python
np.random.seed(0)

w1 = np.random.uniform(-1,1,(2,4))
w2 = np.random.uniform(-1,1,(4,1))

lr = 0.1

for epoch in range(20000):

    hidden_layer = tanh(x @ w1)

    output_layer = tanh(hidden_layer @ w2)

    error = y - output_layer

    d_output = error * tanh_derivative(output_layer)

    d_hidden = (d_output @ w2.T) * tanh_derivative(hidden_layer)

    w2 += hidden_layer.T @ d_output * lr

    w1 += x.T @ d_hidden * lr

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {np.mean(np.abs(error))}")

print("\nFinal output using tanh:")
print(output_layer.round(3))
```

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)
```

```python
np.random.seed(0)

w1 = np.random.uniform(-1,1,(2,4))
w2 = np.random.uniform(-1,1,(4,1))

lr = 0.01

for epoch in range(20000):

    hidden_input = x @ w1
    hidden_layer = relu(hidden_input)

    output_input = hidden_layer @ w2
    output_layer = sigmoid(output_input)

    error = y - output_layer

    d_output = error * sigmoid_derivative(output_layer)

    d_hidden = (d_output @ w2.T) * relu_derivative(hidden_layer)

    w2 += hidden_layer.T @ d_output * lr

    w1 += x.T @ d_hidden * lr

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {np.mean(np.abs(error))}")

print("\nFinal output using ReLU:")
print(output_layer.round(3))
```

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)
```

```python
y_softmax = np.array([
    [1,0],
    [0,1],
    [0,1],
    [1,0]
])
```

```python
np.random.seed(0)

w1 = np.random.uniform(-1,1,(2,4))
w2 = np.random.uniform(-1,1,(4,2))

lr = 0.1

for epoch in range(20000):

    hidden_layer = sigmoid(x @ w1)

    output_layer = softmax(hidden_layer @ w2)

    error = y_softmax - output_layer

    d_output = error

    d_hidden = (d_output @ w2.T) * sigmoid_derivative(hidden_layer)

    w2 += hidden_layer.T @ d_output * lr

    w1 += x.T @ d_hidden * lr

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {np.mean(np.abs(error))}")

print("\nFinal output using Softmax:")
print(output_layer.round(3))
```
